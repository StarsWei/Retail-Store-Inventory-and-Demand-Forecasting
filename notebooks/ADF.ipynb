# 平稳性检验
ADF检验（Augmented Dickey-Fuller检验，增广迪基-富勒检验）  
ADF检验是DF检验的扩展  
DF检验（Dickey-Fuller检验），由Dickey和Fuller于1979年，在论文《Distribution of the Estimators for Autoregressive Time Series with a Unit Root》中提出，是单位根检验的一种方法  
ADF检验由Dickey和Said于1984年在论文《Testing for Unit Roots in Autoregressive Moving Average Models of Unknown Order》中提出  
DF只能处理一阶自相关序列，而ADF可以处理高阶自相关序列

```python
import pandas as pd
from statsmodels.tsa.stattools import adfuller
```

```python
df = pd.read_csv('sales_data.csv', parse_dates=['Date'])
```

```python
ts = df[(df["Store ID"] == "S001") & (df["Product ID"] == "P0001")].set_index("Date")["Units Sold"]
```

```python
adf_result = adfuller(ts)
print(type(adf_result))
print(len(adf_result))
```

```python
print("adf: ", adf_result[0])
print("pvalue: ", adf_result[1])
print("usedlag: ", adf_result[2])
print("nobs: ", adf_result[3])
print("critical values: ", adf_result[4])
print("icbest: ", adf_result[5])
```

```python
adf_stat = adf_result[0]
p_value = adf_result[1]
critical_values = adf_result[4]

# 判断是否平稳
if adf_stat < critical_values['5%'] and p_value < 0.05:
    print("序列是平稳的（拒绝单位根假设）")
else:
    print("序列可能非平稳（无法拒绝单位根假设）")
```
