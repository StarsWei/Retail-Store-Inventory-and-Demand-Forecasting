# 目录树可视化工具
- 从指定目录开始，递归遍历并打印出项目的树状文件结构，自动忽略 .git 和 .ipynb_checkpoints 文件夹。
- 使用缩进与分支符号（├──、└──）展示层级关系
- 仅使用两种图标区分：
    - 📂 文件夹
    - 📄 文件
- 可选参数控制最大递归深度，避免输出过长

```python
import os

IGNORE_DIRS = {'.git', '.ipynb_checkpoints'}

FOLDER_ICON = "📂"
FILE_ICON = "📄"

def print_tree(root_path: str, max_depth: int = None, prefix: str = ""):
    """
    打印目录树结构
    """
    root_path = os.path.abspath(root_path)
    if not os.path.exists(root_path):
        print(f"[ERROR] 路径不存在: {root_path}")
        return
    
    def _walk(current_path, depth, prefix):
        if max_depth is not None and depth > max_depth:
            return
        try:
            entries = sorted(os.listdir(current_path))
        except PermissionError:
            print(prefix + " [权限不足]")
            return
        entries = [e for e in entries if e not in IGNORE_DIRS]
        
        for i, entry in enumerate(entries):
            path = os.path.join(current_path, entry)
            connector = "└── " if i == len(entries) - 1 else "├── "
            icon = FOLDER_ICON if os.path.isdir(path) else FILE_ICON
            print(prefix + connector + f"{icon} {entry}")
            if os.path.isdir(path):
                extension = "    " if i == len(entries) - 1 else "│   "
                _walk(path, depth + 1, prefix + extension)
    
    # 打印根目录
    print(f"{FOLDER_ICON} {os.path.basename(root_path)}/")
    _walk(root_path, 1, "")
```

```python
# 打印上一级目录
print_tree("..")
```
