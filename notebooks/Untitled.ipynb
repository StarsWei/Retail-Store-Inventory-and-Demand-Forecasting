```python
import pandas as pd
import matplotlib.pyplot as plt

# 假设你已经有一个 DataFrame df，包含以下列：
# 'date' (日期: 2022-01-01 ~ 2022-12-31)
# 'promo_coverage' (促销覆盖率, 0~1 或百分比)
# 'sales' (每日销量)

# 示例：如果你还没有数据，可以用下面的代码生成模拟数据
# -----------------------------------------
import numpy as np
date_range = pd.date_range("2022-01-01", "2022-12-31", freq="D")
np.random.seed(42)
df = pd.DataFrame({
    "date": date_range,
    "promo_coverage": np.random.uniform(0.1, 0.8, len(date_range)),  # 模拟促销覆盖率
    "sales": np.random.randint(50, 300, len(date_range))             # 模拟销量
})
# -----------------------------------------

# 设置日期为索引
df.set_index("date", inplace=True)

# 绘图
fig, ax1 = plt.subplots(figsize=(14,6))

# 左轴：促销覆盖率
ax1.plot(df.index, df["promo_coverage"], color="tab:blue", label="促销覆盖率")
ax1.set_ylabel("促销覆盖率", color="tab:blue")
ax1.tick_params(axis="y", labelcolor="tab:blue")

# 右轴：销量
ax2 = ax1.twinx()
ax2.plot(df.index, df["sales"], color="tab:red", label="销量")
ax2.set_ylabel("销量", color="tab:red")
ax2.tick_params(axis="y", labelcolor="tab:red")

# 标题与图例
fig.suptitle("门店S001每日促销覆盖率与销量对比 (2022)", fontsize=16)
fig.legend(loc="upper right", bbox_to_anchor=(0.9, 0.9))

plt.show()
```
